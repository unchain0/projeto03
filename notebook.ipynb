# Visualização de dados

```python
def plot_histogram(column: str, df: pd.DataFrame) -> None:
    plot = px.histogram(
        df,
        x=column,
        y="preco",
        text_auto=True,
        title=f"Faturamento por {column}",
        color="forma_pagamento"
    )
    plot.update_layout(
        xaxis_title=column.title(),
        yaxis_title="Total de vendas",
        title={
            'text': f"Faturamento por {column}",
            'y': 0.9,
            'x': 0.5,
            'xanchor': 'center',
            'yanchor': 'top'
        }
    )
    plot.show()
    plot.write_html(DATA_DIR / f"faturamento-por-{column}.html")
```

```python
columns = ["loja", "cidade", "estado", "tamanho", "local_consumo"]

for column in columns:
    plot_histogram(column, df)
```

**Gráfico de Barras por Tipo de Consumo e Tamanho do Produto**

Comparar a quantidade de pedidos por diferentes tamanhos de produto e tipo de consumo (no local vs delivery).

```python
df_grouped = df.groupby(["tamanho", "local_consumo"]).size().reset_index(name="count")

fig = px.bar(
    df_grouped,
    x="tamanho",
    y="count",
    color="local_consumo",
    title="Local de Consumo x Tamanho do Copo",
)
fig.update_layout(
    xaxis_title="Tamanho do copo",
    yaxis_title="Quantidade de pedidos",
    title={
        'text': "Local de Consumo x Tamanho do Copo",
        'y': 0.9,
        'x': 0.5,
        'xanchor': 'center',
        'yanchor': 'top'
    }
)

fig.show()
```

A quantidade entre o local de consumo com os tamanhos dos copos está equilibrada.

Porém, é possível notar que, existe uma diferença considerável referente ao tamanho do copo com a quantidade que ele é pedido.
